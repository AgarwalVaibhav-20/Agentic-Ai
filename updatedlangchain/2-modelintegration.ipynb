Batch
Batching a collection of independent requests to a model can signify improve performance and reeduce cost , as the processing can be dont parelle:

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

resposnes = model.batch([
    "Why do parrots have colourfuull feather",
    "How do airplanes fly?",
    "What i quantum computing?"
],
config={
    'max_concurrency':5,
})
for response in resposnes:
    print(response)
```
